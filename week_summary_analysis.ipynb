```python
import pandas as pd
```

```python
dta = pd.read_pickle('finra_weekly_summary_2026.pkl')
```

```python
last = dta[(dta['tierIdentifier'] == 'T1') & (dta['summaryStartDate'] == '2026-03-30') & (dta['summaryTypeCode'] == 'ATS_W_SMBL_FIRM')]
```

```python
last[last['issueSymbolIdentifier'] == 'AAPL']['totalWeeklyShareQuantity'].sum()
```

```python
breakdown = last.groupby('MPID')['totalWeeklyShareQuantity'].sum().sort_values(ascending=False).head(50) / 1e6
```

```python
import plotly.express as px

# Filter for top venues (top 10 by total share quantity)
top_venues = breakdown.head(15).reset_index()
top_venues.columns = ['MPID', 'totalWeeklyShareQuantity']

fig = px.pie(top_venues, values='totalWeeklyShareQuantity', names='MPID', 
             title='Top 10 MPIDs by Total Weekly Share Quantity',
             color_discrete_sequence=px.colors.qualitative.Set3,
             hole=0.4)
fig.update_traces(textinfo='percent+label')
fig.show()
```
